```python
import numpy as np
import matplotlib.pyplot as plt

def derivative_first_principles(f, x, h=1e-7):
    """Calculate derivative using first principles"""
    return (f(x + h) - f(x)) / h

# Example: f(x) = x²
def f(x):
    return x**2

# True derivative: f'(x) = 2x
def f_prime_true(x):
    return 2*x

# Test at x = 3
x_test = 3
approx_derivative = derivative_first_principles(f, x_test)
true_derivative = f_prime_true(x_test)

print(f"Approximate derivative at x=3: {approx_derivative:.6f}")
print(f"True derivative at x=3: {true_derivative}")
print(f"Error: {abs(approx_derivative - true_derivative):.8f}")
```

```python
# Numerical derivative using first principles
def derivative(f, x, h=1e-5):
    return (f(x + h) - f(x)) / h

f = lambda x: x**2
print("Approx. derivative at x=3:", derivative(f, 3))  # Should be close to 6
```

#### Usage of derivatives in Gradient Descent

```python
# Gradient descent update rule
def gradient_descent_step(loss_function, weights, learning_rate, h=1e-7):
    """One step of gradient descent using numerical differentiation"""
    gradient = (loss_function(weights + h) - loss_function(weights)) / h
    return weights - learning_rate * gradient

# Example: Simple linear regression
def mse_loss(w, X, y):
    """Mean squared error loss"""
    predictions = X * w
    return np.mean((predictions - y)**2)

# Generate sample data
np.random.seed(42)
X = np.random.randn(100)
y = 2 * X + 1 + 0.1 * np.random.randn(100)  # y = 2x + 1 + noise

# Training loop
weight = 0.0
learning_rate = 0.01

for epoch in range(100):
    # Calculate gradient using first principles
    gradient = derivative_first_principles(lambda w: mse_loss(w, X, y), weight)
    weight -= learning_rate * gradient
    
    if epoch % 20 == 0:
        loss = mse_loss(weight, X, y)
        print(f"Epoch {epoch}: Weight = {weight:.4f}, Loss = {loss:.4f}")
```

### exponential growth vs exponential growth vs sigmoid vs softmax

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Exponential Growth and Decay
x = np.linspace(-3, 3, 100)
exp_growth = np.exp(x)
exp_decay = np.exp(-x)

# 2. Sigmoid Function
sigmoid = 1 / (1 + np.exp(-x))

# Plot Exponential Functions
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.plot(x, exp_growth, label="e^x", color='green')
plt.title("Exponential Growth")
plt.grid(True)
plt.legend()

plt.subplot(2, 2, 2)
plt.plot(x, exp_decay, label="e^(-x)", color='red')
plt.title("Exponential Decay")
plt.grid(True)
plt.legend()

# Plot Sigmoid
plt.subplot(2, 2, 3)
plt.plot(x, sigmoid, label="Sigmoid", color='blue')
plt.title("Sigmoid Function")
plt.grid(True)
plt.legend()

# Plot Softmax Example
scores = np.array([1.0, 2.0, 3.0])
exp_scores = np.exp(scores)
softmax = exp_scores / np.sum(exp_scores)

plt.subplot(2, 2, 4)
plt.bar(['Class A', 'Class B', 'Class C'], softmax, color='purple')
plt.title("Softmax Output (Probabilities)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

#### softmax function

```python
def softmax(scores):
    exp_scores = np.exp(scores - np.max(scores))  # For stability
    return exp_scores / np.sum(exp_scores)

scores = np.array([2.0, 1.0, 0.1])
probs = softmax(scores)
print("Softmax probabilities:", probs)
print("Sum of probabilities:", np.sum(probs))  # Should be 1.0
# plt.subplot(2, 2, 4)
plt.bar(['Class A', 'Class B', 'Class C'], probs, color='purple')
plt.title("Softmax Output (Probabilities)")
plt.grid(True)
```

#### Exponential Learning Rate Decay (Training Example)

```python
initial_lr = 0.1
decay_rate = 0.05
epochs = np.arange(0, 50)
lr = initial_lr * np.exp(-decay_rate * epochs)

plt.plot(epochs, lr, label="Learning Rate")
plt.title("Exponential Decay of Learning Rate")
plt.xlabel("Epochs")
plt.ylabel("Learning Rate")
plt.grid(True)
plt.legend()
plt.show()
```
