```python
import autograd.numpy as anp
from autograd import jacobian

# Vector-valued function: R² → R²
def f(x):
    return anp.array([
        x[0]**2 + x[1],
        anp.sin(x[0] * x[1])
    ])

# Compute Jacobian
jacobian_f = jacobian(f)

# Evaluate at x = [1.0, 2.0]
x_val = anp.array([1.0, 2.0])
J = jacobian_f(x_val)

print("Jacobian Matrix at x = [1, 2]:")
print(J)
```

### Interpretation:
- Row 1: Gradients of f₁ = x² + y → [2x, 1]
- Row 2: Gradients of f₂ = sin(xy) → [y·cos(xy), x·cos(xy)]

```python
import autograd.numpy as anp
from autograd import hessian

# Define scalar-valued function
def f(x):
    return x[0]**2 + 3*x[0]*x[1] + 2*x[1]**2

# Compute Hessian
hessian_f = hessian(f)

# Evaluate at a point
x_val = anp.array([1.0, 2.0])
H = hessian_f(x_val)

print("Hessian matrix at x = [1, 2]:")
print(H)
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Create grid
x = anp.linspace(-3, 3, 100)
y = anp.linspace(-3, 3, 100)
X, Y = anp.meshgrid(x, y)
Z = X**2 + 3*X*Y + 2*Y**2

# Plot 3D surface
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)
ax.set_title("Surface: $f(x, y) = x^2 + 3xy + 2y^2$")
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('f(x, y)')
plt.show()
```
