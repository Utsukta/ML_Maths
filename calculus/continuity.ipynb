```python
# Re-import libraries due to state reset
import numpy as np
import matplotlib.pyplot as plt

# Define learning rate and iterations
lr = 0.1
iterations = 25

# Define the continuous loss function and its gradient
def L1(x):
    return (x - 3)**2

def grad_L1(x):
    return 2 * (x - 3)

# Define the discontinuous loss function (discontinuous at x = 2)
def L2(x):
    return np.where(x == 2, 10, (x - 3)**2)

def grad_L2(x):
    # Gradient is undefined at x = 2 due to discontinuity,
    # we simulate a subgradient approach or skip the update at x = 2
    grad = np.where(x == 2, 0.0, 2 * (x - 3))
    return grad

# Initialize
x1_vals = [0.0]
x2_vals = [0.0]

# Perform gradient descent for both loss functions
for _ in range(iterations):
    x1_vals.append(x1_vals[-1] - lr * grad_L1(x1_vals[-1]))
    x2_vals.append(x2_vals[-1] - lr * grad_L2(x2_vals[-1]))

x1_vals = np.array(x1_vals)
x2_vals = np.array(x2_vals)
loss1 = L1(x1_vals)
loss2 = L2(x2_vals)

# Plot
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(x1_vals, loss1, marker='o', label="Continuous Loss")
plt.title("Gradient Descent on Continuous Function")
plt.xlabel("x")
plt.ylabel("Loss")
plt.grid(True)
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(x2_vals, loss2, marker='x', color='red', label="Discontinuous Loss")
plt.title("Gradient Descent on Discontinuous Function")
plt.xlabel("x")
plt.ylabel("Loss")
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Simulated input values
x = np.linspace(-5, 5, 100)

# Define activation functions
def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

# Compute outputs
relu_vals = relu(x)
sigmoid_vals = sigmoid(x)
tanh_vals = tanh(x)
leaky_relu_vals = leaky_relu(x)

# Plot the results
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.plot(x, relu_vals, label='ReLU', color='blue')
plt.title('ReLU Activation')
plt.grid(True)
plt.legend()

plt.subplot(2, 2, 2)
plt.plot(x, sigmoid_vals, label='Sigmoid', color='green')
plt.title('Sigmoid Activation')
plt.grid(True)
plt.legend()

plt.subplot(2, 2, 3)
plt.plot(x, tanh_vals, label='Tanh', color='purple')
plt.title('Tanh Activation')
plt.grid(True)
plt.legend()

plt.subplot(2, 2, 4)
plt.plot(x, leaky_relu_vals, label='Leaky ReLU', color='orange')
plt.title('Leaky ReLU Activation')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```
