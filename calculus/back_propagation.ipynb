```python
import numpy as np

# Forward Pass
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x = 0.5
w1 = 0.4
w2 = 0.6

# Layer 1
a1 = w1 * x
h1 = sigmoid(a1)

# Layer 2
a2 = w2 * h1
y = sigmoid(a2)

# Backward Pass using Chain Rule
dy_da2 = y * (1 - y)
da2_dw2 = h1

# ∂y/∂w2
dy_dw2 = dy_da2 * da2_dw2

print("dy/dw2 =", dy_dw2)
```

```python
import torch
import torch.nn.functional as F

# Inputs
x1 = torch.tensor(1.0)
x2 = torch.tensor(2.0)

# Weights and biases for hidden layer
W11 = torch.tensor(0.1, requires_grad=True)
W21 = torch.tensor(0.2, requires_grad=True)
b1 = torch.tensor(0.5, requires_grad=True)

W12 = torch.tensor(0.3, requires_grad=True)
W22 = torch.tensor(0.4, requires_grad=True)
b2 = torch.tensor(0.6, requires_grad=True)

# Weights and bias for output layer
W13 = torch.tensor(0.6, requires_grad=True)
W23 = torch.tensor(0.7, requires_grad=True)
b3 = torch.tensor(0.8, requires_grad=True)

# Forward pass
z1 = x1 * W11 + x2 * W21 + b1
h1 = torch.sigmoid(z1)

z2 = x1 * W12 + x2 * W22 + b2
h2 = torch.sigmoid(z2)

z_out = h1 * W13 + h2 * W23 + b3
y_pred = torch.sigmoid(z_out)

# Target label
target = torch.tensor(1.0)

# Loss
loss = F.binary_cross_entropy(y_pred, target)
print(f'loss is {loss}')

# Backward pass
loss.backward()

# Print the gradients
print(f"dL/dW11 = {W11.grad.item():.6f}")
print(f"dL/db1  = {b1.grad.item():.6f}")
print(f"dL/dW13 = {W13.grad.item():.6f}")
```

```python
def sigmoid(x):
    s = 1 / (1 + np.exp(-x))
    return s, s * (1 - s)

def relu(x):
    return np.maximum(0, x), (x > 0).astype(float)

x = np.array([-2.0, 0.0, 2.0])
_, grad_sigmoid = sigmoid(x)
_, grad_relu = relu(x)

print("Sigmoid Derivative:", grad_sigmoid)
print("ReLU Derivative:", grad_relu)
```
