#### Finding citicial points

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize_scalar
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
import seaborn as sns

# Set style for better plots
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")

# 1. CRITICAL POINTS AND DERIVATIVE TESTS

def example_function(x):
    """Example function: f(x) = x^4 - 4x^3 + 4x^2 + 1"""
    return x**4 - 4*x**3 + 4*x**2 + 1

def first_derivative(x):
    """First derivative: f'(x) = 4x^3 - 12x^2 + 8x"""
    return 4*x**3 - 12*x**2 + 8*x

def second_derivative(x):
    """Second derivative: f''(x) = 12x^2 - 24x + 8"""
    return 12*x**2 - 24*x + 8

def find_critical_points():
    """Find and analyze critical points"""
    print("=== CRITICAL POINTS ANALYSIS ===")
    
    critical_points = [0, 1, 2]
    
    print("Critical points found at x =", critical_points)
    
    # Apply second derivative test
    for cp in critical_points:
        f_value = example_function(cp)
        f_prime = first_derivative(cp)
        f_double_prime = second_derivative(cp)
        
        print(f"\nAt x = {cp}:")
        print(f"  f({cp}) = {f_value:.3f}")
        print(f"  f'({cp}) = {f_prime:.3f}")
        print(f"  f''({cp}) = {f_double_prime:.3f}")
        
        if f_double_prime > 0:
            print(f"  → Local MINIMUM (f''({cp}) > 0)")
        elif f_double_prime < 0:
            print(f"  → Local MAXIMUM (f''({cp}) < 0)")
        else:
            print(f"  → Test inconclusive (f''({cp}) = 0)")
    
    return critical_points

def plot_function_analysis():
    """Visualize function and its critical points"""
    x = np.linspace(-0.5, 3, 1000)
    y = example_function(x)
    y_prime = first_derivative(x)
    y_double_prime = second_derivative(x)
    
    critical_points = [0, 1, 2]
    
    fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(15, 10))
    
    # Original function
    ax1.plot(x, y, 'b-', linewidth=2, label='f(x) = x⁴ - 4x³ + 4x² + 1')
    for cp in critical_points:
        ax1.plot(cp, example_function(cp), 'ro', markersize=8)
        ax1.annotate(f'({cp}, {example_function(cp):.1f})', 
                    (cp, example_function(cp)), xytext=(5, 5), 
                    textcoords='offset points')
    ax1.set_title('Original Function f(x)')
    ax1.grid(True, alpha=0.3)
    ax1.legend()
    
    # First derivative
    ax2.plot(x, y_prime, 'g-', linewidth=2, label="f'(x) = 4x³ - 12x² + 8x")
    ax2.axhline(y=0, color='k', linestyle='--', alpha=0.5)
    for cp in critical_points:
        ax2.plot(cp, 0, 'ro', markersize=8)
    ax2.set_title('First Derivative f\'(x)')
    ax2.grid(True, alpha=0.3)
    ax2.legend()
    
    # Second derivative
    ax3.plot(x, y_double_prime, 'r-', linewidth=2, label="f''(x) = 12x² - 24x + 8")
    ax3.axhline(y=0, color='k', linestyle='--', alpha=0.5)
    for cp in critical_points:
        ax3.plot(cp, second_derivative(cp), 'ro', markersize=8)
    ax3.set_title('Second Derivative f\'\'(x)')
    ax3.grid(True, alpha=0.3)
    ax3.legend()
    
    # All together
    ax4.plot(x, y, 'b-', linewidth=2, label='f(x)')
    ax4.plot(x, y_prime, 'g-', linewidth=2, label="f'(x)")
    ax4.plot(x, y_double_prime, 'r-', linewidth=2, label="f''(x)")
    for cp in critical_points:
        ax4.plot(cp, example_function(cp), 'ko', markersize=8)
    ax4.set_title('Function and Its Derivatives')
    ax4.grid(True, alpha=0.3)
    ax4.legend()
    
    plt.tight_layout()
    plt.show()
```

```python
find_critical_points()
```

```python
plot_function_analysis()
```
