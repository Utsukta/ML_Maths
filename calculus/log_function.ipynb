#### basic logarithmic function

```python
import numpy as np

print(np.log(1))         # 0
print(np.log(np.e))      # 1
print(np.log(100))       # Natural log of 100
print(np.log10(100))     # Base-10 log of 100
```

#### Plot Log vs Exponential

```python
import matplotlib.pyplot as plt

x = np.linspace(0.1, 10, 100)
log_y = np.log(x)
exp_y = np.exp(x)

plt.figure(figsize=(10, 5))
plt.plot(x, log_y, label="log(x)", color='blue')
plt.plot(x, exp_y, label="exp(x)", color='green')
plt.title("Logarithmic vs Exponential Functions")
plt.grid(True)
plt.legend()
plt.show()
```

#### Cross-Entropy Loss Demo

```python
def cross_entropy(y_true, y_pred):
    return -np.sum(y_true * np.log(y_pred))

y_true = np.array([0, 1, 0])  # True class is class 1
y_pred_good = np.array([0.01, 0.98, 0.01])
y_pred_bad = np.array([0.6, 0.2, 0.2])

print("Good Prediction Loss:", cross_entropy(y_true, y_pred_good))  # Low loss
print("Bad Prediction Loss:", cross_entropy(y_true, y_pred_bad))    # High loss
```

#### KL-Divergence


```python
def kl_divergence(p, q):
    p = np.array(p)
    q = np.array(q)
    return np.sum(p * np.log(p / (q + 1e-15)))  # small epsilon to avoid log(0)

P = [0.1, 0.8, 0.1]  # True distribution
Q_good = [0.1, 0.8, 0.1]  # Perfect prediction
Q_bad = [0.3, 0.4, 0.3]   # Bad prediction

print("KL divergence (good):", kl_divergence(P, Q_good))  # ~0.0
print("KL divergence (bad):", kl_divergence(P, Q_bad))    # > 0
```
