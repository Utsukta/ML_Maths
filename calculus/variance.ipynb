```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set random seed for reproducibility
np.random.seed(42)

# Simulated "true" function
x = np.linspace(0, 10, 100)
true_y = np.sin(x)

# Simulate high variance model predictions (e.g., from overfitting models)
high_variance_preds = [np.sin(x) + np.random.normal(scale=0.5, size=len(x)) for _ in range(5)]

# Simulate low variance model predictions (more stable model)
low_variance_preds = [np.sin(x) + np.random.normal(scale=0.1, size=len(x)) for _ in range(5)]

# Plotting
fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)

# High variance
axes[0].plot(x, true_y, label="True Function", color="black", linewidth=2)
for pred in high_variance_preds:
    axes[0].plot(x, pred, alpha=0.6)
axes[0].set_title("🔴 High Variance Model Predictions")
axes[0].legend()

# Low variance
axes[1].plot(x, true_y, label="True Function", color="black", linewidth=2)
for pred in low_variance_preds:
    axes[1].plot(x, pred, alpha=0.6)
axes[1].set_title("🟢 Low Variance Model Predictions")
axes[1].legend()

plt.suptitle("Model Prediction Variance in Machine Learning", fontsize=16)
plt.tight_layout()
plt.show()
```
