```python
from itertools import combinations

features = ['f1', 'f2', 'f3', 'f4']
combo = list(combinations(features, 2))
print("Feature combinations (2 at a time):", combo)
```

```python
from itertools import permutations

tokens = ['ML', 'is', 'fun']
perm = list(permutations(tokens))
print("All sentence permutations:", perm)
```

```python
import matplotlib.pyplot as plt

values = [1, 2, 3, 4, 5, 6]
pmf = [1/6] * 6

plt.stem(values, pmf)
plt.title("PMF of a fair die")
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.show()
```
