```python
# Re-import after kernel reset
import numpy as np
import matplotlib.pyplot as plt

# Data points
x = np.array([1, 2, 3])
y = np.array([2, 4, 5])

# Create design matrix X with a column of ones for the intercept
X = np.column_stack((np.ones_like(x), x))

# Compute theta using the normal equation
theta = np.linalg.inv(X.T @ X) @ X.T @ y

# Compute predicted y values (projection)
y_pred = X @ theta

# Plot actual vs predicted values
plt.figure(figsize=(8, 5))
plt.scatter(x, y, color='blue', label='Actual values (y)')
plt.plot(x, y_pred, color='red', label='Best fit line (ŷ = Xθ)', linewidth=2)

# Show projection lines (errors)
for i in range(len(x)):
    plt.plot([x[i], x[i]], [y[i], y_pred[i]], 'k--', linewidth=1)

plt.title('Normal Equation as Projection (Linear Regression)')
plt.xlabel('Hours Studied (x)')
plt.ylabel('Test Score (y)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show(), theta
```
